```python
import astropy
import healpy as hp
import matplotlib.pyplot as plt 
import numpy as np
import shutil

from astropy.io import fits
```

```python
# exposure_map = hp.fitsfunc.read_map("/Volumes/T7/data/sky_map_creation_data/fermi_background_filtered_gti_cmap.fits", hdu=1, nest=False)

# Get number of bins
with fits.open("/Volumes/T7/data/sky_map_creation_data/fermi_filtered_exposure_map.fits") as hdul:
    
    binned_map_data = hdul[0].data
    num_bins = binned_map_data.shape[0]
    energy_bin_boundaries = binned_map_data[1].data
    
for k in range(num_bins - 1):
    
    new_file = "/Volumes/T7/data/sky_map_creation_data/fermi_filtered_exposure_map_bin_{}.fits".format(k)
    
    shutil.copyfile("/Volumes/T7/data/sky_map_creation_data/fermi_filtered_exposure_map.fits", new_file)
    
for k in range(num_bins - 1):
    
    file_name = "/Volumes/T7/data/sky_map_creation_data/fermi_filtered_exposure_map_bin_{}.fits".format(k)
    
    # Create separate binned data files
    with fits.open(file_name, mode='update') as hdul:
        
        binned_map_data = hdul[0].data
        
        healpix_map = binned_map_data[k]
        
        hdul[0].data = hdul[0].data[k]
        
        print(hdul[0].header)
        
        hdul.flush()
```
